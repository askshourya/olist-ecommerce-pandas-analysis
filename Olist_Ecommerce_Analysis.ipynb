```python
# 48.Delivery Time vs Review Score
# Step 1 — Convert dates
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

orders['order_delivered_customer_date'] = pd.to_datetime(
    orders['order_delivered_customer_date']
)
```

```python
# Step 2 — Calculate delivery days
orders['delivery_days'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
```

```python
orders[['order_purchase_timestamp','order_delivered_customer_date','delivery_days']].head()
```

```python
# Step 3 — Merge with reviews
orders_delivery_reviews = orders.merge(reviews,on='order_id',how='left')
```

```python
# Step 4 — Remove missing delivery days
delivery_reviews = orders_delivery_reviews.dropna(
    subset=['delivery_days', 'review_score']
).copy()
```

```python
# Step 5 — Create delivery groups
delivery_reviews['delivery_group'] = pd.cut(
    delivery_reviews['delivery_days'],
    bins=[-1, 3, 7, 14, float('inf')],
    labels=['0-3 days', '4-7 days', '8-14 days', '15+ days']
)
```

```python
delivery_reviews['delivery_group'].value_counts()
```

```python
# Step 6 — Average review score
delivery_reviews.groupby(
    'delivery_group',
    observed=True
)['review_score'].mean().round(2)
```

```python
# 49.Payment Type vs Review Score
payments_reviews = payments.merge(reviews, on='order_id',how='left')
```

```python
# Average review score by payment type
payments_reviews.groupby('payment_type')['review_score'].mean().round(2)
```

```python
# Number of reviews/orders by payment type
payments_reviews.groupby(
    'payment_type'
)['review_score'].count().sort_values(ascending=False)
```

```python
# Better version — both together
payment_analysis = (
    payments_reviews
    .groupby('payment_type')
    .agg(
        reviews=('review_score', 'count'),
        average_rating=('review_score', 'mean')
    )
    .round(2)
    .sort_values('average_rating', ascending=False)
)

payment_analysis
```

```python
# 50.Final Business Question
# 1. Delivery time
delivery_reviews.groupby(
    'delivery_group',
    observed=True
)['review_score'].mean().round(2)
```

```python
delivery_analysis = (
    delivery_reviews
    .groupby('delivery_group', observed=True)
    .agg(
        reviews=('review_score', 'count'),
        average_rating=('review_score', 'mean')
    )
    .round(2)
)

delivery_analysis
```

```python
# 2.Payment method
payment_analysis
```

```python
# 3. Order status
status_analysis = (
    order_reviews
    .groupby('order_status')
    .agg(
        reviews=('review_score', 'count'),
        average_rating=('review_score', 'mean')
    )
    .round(2)
    .sort_values('average_rating', ascending=False)
)

status_analysis
```

```python
# 4.Product category → Review score
# Step 1 — Orders + Reviews
order_review = orders.merge(
    reviews,
    on='order_id',
    how='inner'
)
```

```python
# Step 2 — Add order items
order_review_items = order_review.merge(
    order_items,
    on='order_id',
    how='inner'
)
```

```python
# Step 3 — Add product information
order_review_products = order_review_items.merge(
    products,
    on='product_id',
    how='left'
)
```

```python
# Step 4 — Average rating by category
category_reviews = (
    order_review_products
    .groupby('product_category_name')
    .agg(
        reviews=('review_score', 'count'),
        average_rating=('review_score', 'mean')
    )
    .round(2)
    .sort_values('average_rating')
)

category_reviews.head(10)
```
